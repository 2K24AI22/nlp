# Task 1: Bag of Words Matrix Construction

## Objective

The objective of this task is to convert customer reviews into numerical vectors using the **Bag of Words (BoW)** technique.

We will use `CountVectorizer` from Scikit-learn to:
- Remove common English stop words.
- Create a vocabulary from the given reviews.
- Convert the text into a Term-Frequency (TF) matrix.
- Display the resulting matrix using a Pandas DataFrame.

## Step 1: Define the Customer Reviews

In this step, we define three customer reviews as our text corpus.

A **corpus** is a collection of text documents that we want to process and analyze.

```python
corpus = [
    "The product performance is amazing and fast",
    "The service was fast and performance was great",
    "Terrible customer service and bad performance"
]
```

## Step 2: Initialize CountVectorizer

`CountVectorizer` converts text documents into numerical vectors based on the frequency of words.

We use `stop_words='english'` to remove common English words such as:

- the
- is
- and
- was

These words usually provide less useful information for text analysis.

The `fit_transform()` method first learns the vocabulary from the corpus and then converts each document into a numerical vector.

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer(stop_words='english')

X = vectorizer.fit_transform(corpus)
```

## Step 3: Extract the Vocabulary

The vocabulary contains all unique words identified by `CountVectorizer` after removing the English stop words.

`get_feature_names_out()` returns the learned vocabulary in alphabetical order.

Each vocabulary word represents one feature/column in the Bag of Words matrix.

```python
vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:")
print(vocabulary)
```

## Step 4: Convert the BoW Matrix into a Pandas DataFrame

The transformed result is stored as a sparse matrix.

For easier visualization and structured presentation, we convert it into a Pandas DataFrame.

Each row represents a document, while each column represents a vocabulary word.

The values represent the number of times a word appears in each document.

```python
import pandas as pd

df = pd.DataFrame(
    X.toarray(),
    columns=vocabulary
)

df
```

## Interpretation

The Bag of Words matrix represents each customer review numerically.

For example, if the value under the `performance` column is `1`, it means that the word "performance" appears once in that document.

A value of `0` means that the particular word does not appear in the document.

Therefore, the text data has been successfully converted into a numerical representation that can be used by Machine Learning algorithms.

# Task 2: Document Search Engine & Relevance Ranking

## Objective

The objective of this task is to build a simple document search engine.

The system will:
- Convert documents into numerical vectors.
- Convert the search query into the same vector space.
- Calculate Cosine Similarity between the query and each document.
- Rank the documents from most relevant to least relevant.

## Step 1: Define Documents and Search Query

We define four documents related to Machine Learning, Deep Learning, NLP, and Data Science.

We also define a search query:

**"machine learning algorithms for data"**

The search engine will determine which document is most relevant to this query.

```python
documents = [
    "Machine learning algorithms analyze structured data effectively",
    "Deep learning and neural networks excel at processing unstructured data",
    "Natural language processing helps computers understand human language",
    "Python is widely used for machine learning and data science"
]

query = ["machine learning algorithms for data"]
```

## Step 2: Convert Documents into Numerical Vectors

We use `CountVectorizer` to convert the documents into numerical vectors.

The `fit_transform()` method learns the vocabulary from the documents and creates a numerical representation based on word frequency.

The same vectorizer must be used for the query so that the query and documents have the same feature space.

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()

doc_vectors = vectorizer.fit_transform(documents)
```

## Step 3: Transform the Search Query

The query is transformed using the already fitted `CountVectorizer`.

We use `transform()` instead of `fit_transform()` because we do not want to create a new vocabulary for the query.

The query must use the same vocabulary as the documents for a valid similarity comparison.

```python
query_vector = vectorizer.transform(query)
```

## Step 4: Calculate Cosine Similarity

Cosine Similarity measures how similar two vectors are based on the angle between them.

In this task, we compare the query vector with every document vector.

A higher Cosine Similarity score means that the document is more relevant to the search query.

A score close to `0` means that the document has little or no similarity with the query.

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_scores = cosine_similarity(
    query_vector,
    doc_vectors
)[0]

similarity_scores
```

## Step 5: Rank Documents

We create a Pandas DataFrame containing the documents and their Cosine Similarity scores.

Finally, we sort the results in descending order.

The document with the highest similarity score is ranked first because it is the most relevant to the query.

```python
results = pd.DataFrame({
    "Document": documents,
    "Cosine Similarity": similarity_scores
})

results = results.sort_values(
    by="Cosine Similarity",
    ascending=False
).reset_index(drop=True)

results
```

## Interpretation

The search engine ranks documents according to their similarity with the query:

**"machine learning algorithms for data"**

The document containing the highest number of relevant query terms receives the highest Cosine Similarity score and is ranked first.

This demonstrates how a simple search engine can use text vectorization and similarity measurement to retrieve relevant documents.

Although this approach is simple and effective for basic search tasks, Bag of Words does not understand word order, context, or deeper semantic meaning.
